Launch the VoiceTut web app (voice gallery, diacritics toolbar, voice cloning with upload or mic, streaming, light/dark mode) and open it in a new tab through Colab's built-in port proxy.

```python
# Start the server in the background (the model downloads on first run, a few minutes)
import time, urllib.request
from voicetut_tts.server import launch

launch(port=7860, block=False)       # returns once the server is up
while True:
    try:
        urllib.request.urlopen("http://localhost:7860/healthz"); break
    except Exception:
        time.sleep(3)
print("Server is up")

from google.colab import output
output.serve_kernel_port_as_window(7860)   # click the link that appears
```

> Click the **`https://localhost:7860/`** link printed above. Colab opens it in a new tab through its proxy. The app keeps running while this notebook is connected.
